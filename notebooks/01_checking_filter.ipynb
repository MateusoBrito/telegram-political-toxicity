```python
from pyspark.sql import SparkSession

# Inicializa a sessão Spark
spark = SparkSession.builder \
    .appName("verificacao_duplicatas") \
    .config("spark.driver.memory", "4g") \
    .config("spark.executor.memory", "4g") \
    .getOrCreate()

# Caminho do diretório
PATH_EN = "/home/students/moliveira/messages_july_en_parquet/"

# Leitura do formato Parquet (sem precisar declarar o schema!)
df_en = spark.read.parquet(PATH_EN)

# Selecionamos 'channel_id' no lugar de 'file_path'
df_pk = df_en.select("id", "channel_id")

print(f"Total de linhas únicas: {df_pk.count()}")
```

```python
df_pk = df_en.select("id", "file_path")
print(f"Total de linhas únicas: {df_pk.count()}")
```

```python
df_apenas_lang = df_en.select("lang")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import desc
df_contagem_idiomas = df_apenas_lang.groupBy("lang").count().orderBy(desc("count"))
df_contagem_idiomas.show(50, truncate=False)
```
