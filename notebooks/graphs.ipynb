```python
df_graph = pd.read_parquet(DATA_DIR, columns = ["id", "user_id", "group_name"])
len(df_graph)
```

```python
# 1. Dimensões Gerais
total_msgs = len(df_graph)
total_grupos = df_graph['group_name'].nunique()

# 2. Análise de Valores Nulos (Mensagens anónimas ou disparos de canais)
nulos_user_id = df_graph['user_id'].isna().sum()
pct_nulos = (nulos_user_id / total_msgs) * 100

# 3. Preparação para o Grafo (remover nulos, pois não formam arestas)
df_users = df_graph.dropna(subset=['user_id'])
total_users = df_users['user_id'].nunique()

# 4. Distribuição de Mensagens por Utilizador
msgs_por_user = df_users['user_id'].value_counts()

# 5. Comportamento de Coautoria (O motor do seu grafo)
# Conta em quantos grupos únicos cada utilizador publicou
grupos_por_user = df_users.groupby('user_id')['group_name'].nunique()

# Identifica os utilizadores "ponte" (que conectam 2 ou mais grupos)
users_multi_grupo = grupos_por_user[grupos_por_user > 1]
pct_multi_grupo = (len(users_multi_grupo) / total_users) * 100

print(f"--- ESTATÍSTICAS DA BASE ---")
print(f"Total de mensagens: {total_msgs:,}")
print(f"Total de grupos únicos: {total_grupos:,}")
print(f"Mensagens sem user_id (NaN): {nulos_user_id:,} ({pct_nulos:.2f}%)\n")

print(f"--- ESTATÍSTICAS DE UTILIZADORES ---")
print(f"Total de utilizadores únicos: {total_users:,}")
print(f"Média de mensagens por utilizador: {msgs_por_user.mean():.2f}")
print(f"Máximo de mensagens por um único utilizador: {msgs_por_user.max():,}\n")

print(f"--- POTENCIAL DO GRAFO (COAUTORIA) ---")
print(f"Utilizadores em apenas 1 grupo: {total_users - len(users_multi_grupo):,}")
print(f"Utilizadores em > 1 grupo (Pontes): {len(users_multi_grupo):,} ({pct_multi_grupo:.2f}%)")
print(f"Máximo de grupos frequentados por um único utilizador: {grupos_por_user.max()}")
```
