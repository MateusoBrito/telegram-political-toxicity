```python
import os
import glob
import pandas as pd

base_path = "/home/students/moliveira/telegram_2024/extracted/"
arquivos = glob.glob(os.path.join(base_path, "*", "20*.tsv.gz"))

from collections import defaultdict
file_map = defaultdict(list)
```

```python
for path in arquivos:
    parts = path.split(os.sep)
    channel_id = parts[-2] 
    file_map[channel_id].append(path)
```

```python
print(f"Número de chats total: {len(file_map)}")
```

```python
from collections import Counter
import multiprocessing

num_cores = multiprocessing.cpu_count()

# Executa o processamento paralelo
with multiprocessing.Pool(processes=num_cores) as pool:
    # Use 'arquivos_filtrados' que criamos no passo anterior
    resultados = pool.map(processar_arquivo, arquivos)

# Agrega os totais por mês
totais_por_mes = Counter()

for mes, contagem in resultados:
    totais_por_mes[mes] += contagem

# Exibe os resultados ordenados por mês
print("Total de mensagens por mês:")
for mes in sorted(totais_por_mes.keys()):
    print(f"{mes}: {totais_por_mes[mes]} mensagens")
```
