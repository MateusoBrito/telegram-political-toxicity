```python
import sys
import os

caminho_raiz = os.path.abspath('..')

if caminho_raiz not in sys.path:
    sys.path.append(caminho_raiz)

from src.classifier.comparison import ResultsComparator
from pathlib import Path
```

```python
"""Diretório em que foram salvos os resultados do classificador"""
BASE_DIR = Path("../reports/classifier_macrotopic_balanced_c")

comparator = ResultsComparator(base_dir = BASE_DIR)
```

```python
results = {
    'distilroberta_lr': comparator.get_model_data("stratified_sample_full", representation = 'all-distilroberta-v1', model_name = 'lr'),
    'minilm_lr': comparator.get_model_data("stratified_sample_full", representation = 'all-MiniLM-L6-v2', model_name = 'lr'),
    'paraphrase_lr': comparator.get_model_data("stratified_sample_full", representation = 'paraphrase-MiniLM-L6-v2', model_name = 'lr'),
    'distilroberta_dt': comparator.get_model_data("stratified_sample_full", representation = 'all-distilroberta-v1', model_name ='dt'),
    'minilm_dt': comparator.get_model_data("stratified_sample_full", representation = 'all-MiniLM-L6-v2', model_name ='dt'),
    'paraphrase_dt': comparator.get_model_data("stratified_sample_full", representation = 'paraphrase-MiniLM-L6-v2', model_name ='dt')
}
```

```python
results
```

```python
comparator.save_statistical_report(results, title_suffix='Triagem de Modelos', output_path=BASE_DIR / 'estatistica.txt')
comparator.plot_pairwise_differences(results, metric='f1', title_suffix='F1-Score Pareado', output_path= BASE_DIR / 'diff_f1.png')
```

```python
results
```

```python
import pandas as pd

data_rows = []

for model_name, metrics in results.items():
  acc = metrics['accuracy']
  f1_macro = metrics['f1_macro']
  f1_weighted = metrics['f1_weighted']

  row = {
      'Model/Method': model_name,
      'Accuracy': f'{acc.mean()*100:.2f} ({acc.std()*100:.2f})',
      'F1-Macro': f'{f1_macro.mean()*100:.2f} ({f1_macro.std()*100:.2f})',
      'F1-Weighted': f'{f1_weighted.mean()*100:.2f} ({f1_weighted.std()*100:.2f})',
  }
  data_rows.append(row)

df_table = pd.DataFrame(data_rows)

# Exibe a tabela em formato Markdown
print(df_table.to_markdown(index=False))

# Se preferir o código pronto em LaTeX para colar no Overleaf:
# print(df_table.to_latex(index=False, escape=False))
```
