```python
df_full = pd.read_parquet("../data/processed/messages_preprocessed")
df_full
```

```python
pk_cols = ["id", "group_name"]

# ────────────────────────────────────────────────────────────────
# 1. VALIDAÇÃO DE CHAVE PRIMÁRIA (PK)
# ────────────────────────────────────────────────────────────────
def validate_pk(df, name):
    print(f"\nValidando PK para: {name}")
    print(f"Total de linhas: {len(df)}")
    
    # Verifica nulos nas colunas de chave
    nulls = df[pk_cols].isnull().sum().sum()
    print(f"Valores nulos nas colunas de chave: {nulls}")
    
    # Verifica duplicatas da combinacao id + group_name
    dupes = df.duplicated(subset=pk_cols).sum()
    print(f"Chaves duplicadas (id + group_name): {dupes}")
    
    is_valid = (nulls == 0) and (dupes == 0)
    if is_valid:
        print(f"-> OK: '{name}' forma uma Chave Primária válida e única!")
    else:
        print(f"-> AVISO: '{name}' POSSUI DUPLICATAS OU NULOS NAS CHAVES!")
    return is_valid

valid_full = validate_pk(df_full, "df_full")
valid_classified = validate_pk(df, "df")
```

```python
print("Cruzando os datasets para validar o alinhamento...")
df_check = pd.merge(
    df,
    df_full,
    on=pk_cols,
    suffixes=('_class', '_full'),
    how='inner'
)

print(f"Total de registros cruzados: {len(df_check)}")

# 2. Verificar se a coluna 'text' bate 100%
if 'text_class' in df_check.columns and 'text_full' in df_check.columns:
    match_text = (df_check['text_class'] == df_check['text_full']).all()
    print(f"-> A coluna 'text' é idêntica em todas as linhas? {match_text}")
    if not match_text:
        diffs = (df_check['text_class'] != df_check['text_full']).sum()
        print(f"   Atenção: Existem {diffs} divergências no 'text'.")

# 3. Verificar se a coluna 'clean_text' bate 100% (caso o df_classified já possua essa coluna)
if 'clean_text_class' in df_check.columns and 'clean_text_full' in df_check.columns:
    match_clean = (df_check['clean_text_class'] == df_check['clean_text_full']).all()
    print(f"-> A coluna 'clean_text' é idêntica em todas as linhas? {match_clean}")
    if not match_clean:
        diffs_clean = (df_check['clean_text_class'] != df_check['clean_text_full']).sum()
        print(f"   Atenção: Existem {diffs_clean} divergências no 'clean_text'.")
else:
    print("-> A coluna 'clean_text' não estava presente em df_classified (será importada agora do df_full).")
```

```python
df_final = pd.merge(
    df,
    df_full[pk_cols + ["clean_text"]],
    on=pk_cols,
    how="inner"
)
```
